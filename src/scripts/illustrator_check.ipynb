```python
import pandas as pd
from difflib import SequenceMatcher

# Load your file
df = pd.read_csv("./data/illustrator_list.csv")
artists = sorted(df['illustrator'].dropna().unique().tolist())

print("🔍 Scanning for anomalies...")
for i in range(len(artists)):
    for j in range(i + 1, len(artists)):
        a, b = artists[i], artists[j]
        ratio = SequenceMatcher(None, a.lower(), b.lower()).ratio()
        
        if a.lower() == b.lower() and a != b:
            print(f"⚠️ Case mismatch: '{a}' | '{b}'")
        elif (a in b or b in a) and len(a) > 5:
            print(f"✂️ Substring overlap: '{a}' | '{b}'")
        elif ratio > 0.88:
            print(f"🤔 Possible typo ({ratio:.2f}): '{a}' | '{b}'")
```
